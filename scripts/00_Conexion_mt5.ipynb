```python
import MetaTrader5 as mt5
import pandas as pd
from datetime import datetime

# Initialize MT5 connection
if not mt5.initialize():
    print("initialize() failed, error code =", mt5.last_error())
    quit()

# Define symbol and timeframe
symbol = "EURUSD"
timeframe = mt5.TIMEFRAME_H1 # Hourly timeframe

# Define the number of bars to retrieve or a specific date range
# Option 1: Get the last N bars
num_bars = 100
rates = mt5.copy_rates_from_pos(symbol, timeframe, 0, num_bars)

# Option 2: Get bars within a specific date range
# from_date = datetime(2024, 1, 1)
# to_date = datetime(2024, 1, 31)
# rates = mt5.copy_rates_range(symbol, timeframe, from_date, to_date)

# Check if data was retrieved successfully
if rates is None:
    print(f"Failed to get rates for {symbol}, error code =", mt5.last_error())
else:
    # Create a pandas DataFrame from the retrieved data
    rates_frame = pd.DataFrame(rates)
    # Convert time in seconds to datetime objects
    rates_frame['time'] = pd.to_datetime(rates_frame['time'], unit='s')
    # Set time as index
    rates_frame.set_index('time', inplace=True)

    # Print the OHLC data
    print(rates_frame[['open', 'high', 'low', 'close']])

# Shutdown MT5 connection
mt5.shutdown()
```
