```python
def obtener_z_score(valor, price, t = 14):
    
    today = dt.date.today()
    df = yf.download(tickers=valor, period=f'{2 * t}d', interval='1d', end = today)
    df = df.reset_index() # Se resetea el índice
    df = df[df['Date'] < pd.to_datetime(today)] # Se elimina el día actual incompleto
    df.columns = ['Date', 'Close', 'High', 'Low', 'Open', 'Volume']
    
    #df = df.tail(t)
    df['MA'] = df['Close'].rolling(window=t).mean()
    df['STD'] = df['Close'].rolling(window=t).std()
    df['Upper'] = df['MA'] + (df['STD'] * 2)
    df['Lower'] = df['MA'] - (df['STD'] * 2)
    
    MA, std = df['MA'].values[-1], df['STD'].values[-1]
    #print(f'Close: {close}, MA: {MA}, STD: {std}')
    #df['Z_score'] = (df['Close'] - df['MA']) / df['STD']
    #display(df)
    
    z_score_actual = float((price - MA) / std)
    print(f'Z-score actual: {z_score_actual}')
    return z_score_actual

```

```python
def generar_max_proporcion(z_score, regla):
    # regla = [min no importa el caso, max a z_score = -2, max a z_score = 0, max a z_score = 2, max no importa el caso]
    # máxima proporción de la cuenta que se va a usar en la operación
    # 1. interpolar

    if (z_score >= 0):
        max_proporcion = regla[2] + ( (z_score - 0) * (regla[1] - regla[2]) / (2 - 0) ) # ecuación de la recta
    else:
        max_proporcion = regla[3] + ( (z_score - (-2)) * (regla[2] - regla[3]) / (0 - (-2)) ) # ecuación de la recta
    
    if max_proporcion < regla[0]:
        max_proporcion = regla[0]
    if max_proporcion > regla[-1]:
        max_proporcion = regla[-1]
    
    return max_proporcion
    
```

```python
def definir_lotes(valor, info_valores, regla):

    if valor in info_valores.keys():
        lote_size = info_valores[valor]['lote']
        apalancamiento = info_valores[valor]['apalancamiento']
        yf_symbol = info_valores[valor]['yfsymbol']
        sl_lp = info_valores[valor]['SL_LP']
        min_lotaje = info_valores[valor]['min_lotaje']
    
    #print(lote, apalancamiento)

    balance = obtener_dinero_cuenta()
    try:
        price = mt5.symbol_info_tick(valor).ask
    except:
        print('Error al obtener precio de', valor)
        return 0, sl_lp
    print('Precio actual:', price)
    z_score = obtener_z_score(yf_symbol, price)

    max_proporcion = generar_max_proporcion(z_score, regla)
    lotes_ejecucion = ((balance * (max_proporcion / 100) / (lote_size * (price - sl_lp))) // min_lotaje) * min_lotaje # al final, división parte entera y, se compra esa cantida de veces, el mínimo
    usd_invertidos = price * lote_size * lotes_ejecucion / apalancamiento
    print(f'Lotes a ejecutar: {lotes_ejecucion} | USD invertidos: {usd_invertidos}')
    return lotes_ejecucion, sl_lp
#definir_lotes('BTCUSD', info_valores)
```

```python
def ejecutar_orden(request, symbol, volume, price):
    if 'sl' in request.keys(): #250902
        request['sl'] = float(round(request['sl'], 0))
        
    if volume == 0:
        print('No se ejecuta por volumen = 0')
        return None
    # send a trading request
    result = mt5.order_send(request)
    # check the execution result
    print("1. order_send(): by {} {} lots at {} with deviation={} points".format(symbol, volume, price, 10));
    #print(result)
    if result.retcode != mt5.TRADE_RETCODE_DONE:
        print("2. order_send failed, retcode={}".format(result.retcode))
        # request the result as a dictionary and display it element by element
        result_dict = result._asdict()
        for field in result_dict.keys():
            print("   {}={}".format(field, result_dict[field]))
            # if this is a trading request structure, display it element by element as well
            if field == "request":
                traderequest_dict = result_dict[field]._asdict()
                for tradereq_filed in traderequest_dict:
                    print("       traderequest: {}={}".format(tradereq_filed, traderequest_dict[tradereq_filed]))
        print("shutdown() and quit")
        mt5.shutdown()
        quit()
    
    return None
```

```python
def cambiar_SL(valor, sl):
  
    # Establish connection to MT5
    if not mt5.initialize():
        print("initialize() failed, error code =", mt5.last_error())
        quit()

    # Get all pending orders
    # Use mt5.orders_get() to retrieve all orders.
    # Filter for pending orders based on their type (e.g., mt5.ORDER_TYPE_BUY_LIMIT, mt5.ORDER_TYPE_SELL_STOP, etc.)
    # A more robust approach would be to check if the order is a pending order.
    # obtener ordenes abiertas actualmente
    actual_positions = mt5.positions_get(symbol = valor)

    verify = True
    if actual_positions is None:
        print("No pending orders found or orders_get() failed.")
    else:
        for position in actual_positions:
            print(position)
            # Hacer cambio de SL
            print(position.ticket)
            #print(position.sl)
            #sys.exit('Cambiar SL')
            print('SL', sl)
            
            request = {
            "action": mt5.TRADE_ACTION_SLTP,
            "position": position.ticket,
            "symbol": valor,
            "sl": sl,
            "magic": 234000, # Example magic number
            "comment": "Modify SL via Python"
            }
            
            result = mt5.order_send(request)
            if result.retcode == mt5.TRADE_RETCODE_DONE:
                print(f"Stop Loss modified successfully for position {position.ticket}.")
            else:
                print(f"Failed to modify Stop Loss: {result.comment}")
                verify = False
        if verify:
            print(f'Cambios exitosos para: {valor}')
    
    return None
```

```python
def cerrar_operaciones_abiertas(valor):
    
    if not mt5.initialize():
        print("initialize() failed, error code =", mt5.last_error())
        quit()
    
    positions = mt5.positions_get(symbol = valor)
    if positions is None:
        print("No positions found, error code =", mt5.last_error())
        mt5.shutdown()
        quit()
    
    for position in positions:
        # Determine the closing action (SELL for BUY positions, BUY for SELL positions)
        if position.type == mt5.ORDER_TYPE_BUY:
            order_type = mt5.ORDER_TYPE_SELL
            price = mt5.symbol_info_tick(position.symbol).bid # Use Bid price for closing BUY
        elif position.type == mt5.ORDER_TYPE_SELL:
            order_type = mt5.ORDER_TYPE_BUY
            price = mt5.symbol_info_tick(position.symbol).ask # Use Ask price for closing SELL
        else:
            continue # Skip if position type is not BUY or SELL
        
        # DEtermina si tiene ganancia o no
        print('IMPORTANTE PARA CONTROL DE CIERRE')
        profit = position.profit
        if profit > 0:
            print(f'La posición {position.ticket} tiene ganancia de {profit}')
        else:
            print(f'La posición {position.ticket} tiene pérdida de {profit}')
            

        # Prepare the close request
        request = {
            "action": mt5.TRADE_ACTION_DEAL,
            "symbol": position.symbol,
            "volume": position.volume,
            "type": order_type,
            "price": price,
            "deviation": 10, # Adjust deviation as needed
            "position": position.ticket,
            "comment": "Closing position by symbol",
            "type_filling": mt5.ORDER_FILLING_FOK, # Or other filling types
        }

        # Send the close request
        result = mt5.order_send(request)
        if result.retcode != mt5.TRADE_RETCODE_DONE:
            print(f"Failed to close position {position.ticket}, retcode={result.retcode}")
        else:
            print(f"Position {position.ticket} closed successfully.")
    
    return None
```
